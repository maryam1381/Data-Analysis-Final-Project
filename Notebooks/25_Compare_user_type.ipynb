## 25. مقایسه مشاور املاک و کاربر شخصی

پرسش:

> آیا مشاوران املاک و کاربران شخصی برای واحدهای مشابه قیمت متفاوتی ثبت می‌کنند؟

حداقل کنترل‌ها:

- شهر و محله
- نوع ملک
- مساحت
- سن بنا
- تعداد اتاق
- ماه
- امکانات اصلی
- رژیم قیمت

خروجی باید شامل:

- تفاوت خام
- تفاوت کنترل‌شده
- اندازه نمونه
- اندازه اثر
- عدم قطعیت
- محدودیت انتخاب غیرتصادفی نوع کاربر

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_feather("../Outputs/21_df.feather")
```

```python
# ==============================================
# مقایسه مشاور املاک و کاربر شخصی بر اساس قیمت هر متر مربع
# ==============================================

print("=" * 80)
print("📊 مقایسه قیمت هر متر مربع: مشاور املاک vs کاربر شخصی")
print("=" * 80)

# ==============================================
# 1. رژیم فروش (sell) - sale_price_per_sqm
# ==============================================

print("\n" + "="*80)
print(" رژیم فروش (sell) - قیمت هر متر مربع")
print("="*80)

sell_df = df[
    (df['price_regime'] == 'sell') &
    (df['user_type'].isin(['مشاور املاک', 'شخصی'])) &
    (df['sale_price_per_sqm'] > 0)& (df["price_status"]=="valid")
].copy()

if len(sell_df) > 0:
    agent = sell_df[sell_df['user_type'] == 'مشاور املاک']
    personal = sell_df[sell_df['user_type'] == 'شخصی']
    
    # تفاوت خام
    mean_agent = agent['sale_price_per_sqm'].mean()
    mean_personal = personal['sale_price_per_sqm'].mean()
    median_agent = agent['sale_price_per_sqm'].median()
    median_personal = personal['sale_price_per_sqm'].median()

    raw_diff = mean_agent - mean_personal
    raw_pct = (raw_diff / mean_personal) * 100

    raw_diff1 = median_agent - median_personal
    raw_pct1 = (raw_diff1 / median_personal) * 100
    
    print(f"تعداد مشاوران: {len(agent):,}, میانگین: {mean_agent:,.0f}, میانه: {median_agent:,.0f}")
    print(f"تعداد شخصی: {len(personal):,}, میانگین: {mean_personal:,.0f}, میانه: {median_personal:,.0f}")
    print(f"تفاوت خام میانگین: {raw_diff:,.0f} ({raw_pct:.1f}%)")
    print(f"تفاوت خام میانه: {raw_diff1:,.0f} ({raw_pct1:.1f}%)")
    

# ==============================================
# 2. رژیم فقط اجاره (rent_only) - equivalent_monthly_rent
# ==============================================

print("\n" + "="*80)
print(" رژیم اجاره کامل (rent_only) - اجاره ماهانه به ازای هر متر")
print("="*80)

rent_df = df[
    (df['price_regime'] == 'rent_only') &
    (df['user_type'].isin(['مشاور املاک', 'شخصی'])) &
    (df['equivalent_monthly_rent'] > 0)& (df["price_status"]=="valid")
].copy()

if len(rent_df) > 0:
    agent = rent_df[rent_df['user_type'] == 'مشاور املاک']
    personal = rent_df[rent_df['user_type'] == 'شخصی']
    
    mean_agent = agent['equivalent_monthly_rent'].mean()
    mean_personal = personal['equivalent_monthly_rent'].mean()

    median_agent = agent['equivalent_monthly_rent'].median()
    median_personal = personal['equivalent_monthly_rent'].median()

    raw_diff = mean_agent - mean_personal
    raw_pct = (raw_diff / mean_personal) * 100

    raw_diff1 = median_agent - median_personal
    raw_pct1 = (raw_diff1 / median_personal) * 100
    
    print(f"تعداد مشاوران: {len(agent):,}, میانگین: {mean_agent:,.0f}, میانه: {median_agent:,.0f}")
    print(f"تعداد شخصی: {len(personal):,}, میانگین: {mean_personal:,.0f}, میانه: {median_personal:,.0f}")
    print(f"تفاوت خام میانگین: {raw_diff:,.0f} ({raw_pct:.1f}%)")
    print(f"تفاوت خام میانه: {raw_diff1:,.0f} ({raw_pct1:.1f}%)")
    
    

# ==============================================
# 3. رژیم فقط رهن (mortgage_only) - equivalent_deposit
# ==============================================

print("\n" + "="*80)
print(" رژیم رهن کامل (mortgage_only) - مبلغ رهن به ازای هر متر")
print("="*80)

mortgage_df = df[
    (df['price_regime'] == 'mortgage_only') &
    (df['user_type'].isin(['مشاور املاک', 'شخصی'])) &
    (df['equivalent_deposit'] > 0)& (df["price_status"]=="valid")
].copy()

if len(mortgage_df) > 0:
    agent = mortgage_df[mortgage_df['user_type'] == 'مشاور املاک']
    personal = mortgage_df[mortgage_df['user_type'] == 'شخصی']
    
    mean_agent = agent['equivalent_deposit'].mean()
    mean_personal = personal['equivalent_deposit'].mean()

    median_agent = agent['equivalent_deposit'].median()
    median_personal = personal['equivalent_deposit'].median()

    raw_diff = mean_agent - mean_personal
    raw_pct = (raw_diff / mean_personal) * 100

    raw_diff1 = median_agent - median_personal
    raw_pct1 = (raw_diff1 / median_personal) * 100
    
    print(f"تعداد مشاوران: {len(agent):,}, میانگین: {mean_agent:,.0f}, میانه: {median_agent:,.0f}")
    print(f"تعداد شخصی: {len(personal):,}, میانگین: {mean_personal:,.0f}, میانه: {median_personal:,.0f}")
    print(f"تفاوت خام میانگین: {raw_diff:,.0f} ({raw_pct:.1f}%)")
    print(f"تفاوت خام میانه: {raw_diff1:,.0f} ({raw_pct1:.1f}%)")

```
