## 27. بخش‌بندی بازار

آگهی‌ها یا محله‌ها را می‌توان براساس موارد زیر بخش‌بندی کرد:

- قیمت
- مساحت
- امکانات
- نوع ملک
- موقعیت
- سن بنا
- رژیم قیمت

بخش‌ها باید قابل تفسیر باشند؛ برای مثال:

- اقتصادی
- میان‌رده
- لوکس
- واحد کوچک شهری
- خانوادگی
- زمین سرمایه‌ای

### انتظار مدرس

- Feature scaling مستند باشد.
- تعداد Clusterها توجیه شود.
- Missingها پیش از Clustering مدیریت شوند.
- Segmentها فقط شماره نباشند و Profile شوند.
- پایداری Segmentها بررسی شود.
- از مخلوط‌کردن رژیم‌های قیمت ناسازگار خودداری شود.
